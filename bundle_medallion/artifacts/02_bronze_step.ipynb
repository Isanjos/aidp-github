# Medallion demo -- bronze 02
Idempotent demo pipeline step.

```python
from pyspark.sql import functions as F
spark.sql("CREATE SCHEMA IF NOT EXISTS demo_bronze")
rows = [(201, "item_2_a", 20.0), (202, "item_2_b", 40.0)]
df = spark.createDataFrame(rows, ["record_id", "item", "amount"])
df = df.withColumn("source_step", F.lit(2)).withColumn("ingested_at", F.current_timestamp())
df.write.mode("overwrite").format("delta").saveAsTable("demo_bronze.step_02_raw")
print("bronze step 02 complete")
```
