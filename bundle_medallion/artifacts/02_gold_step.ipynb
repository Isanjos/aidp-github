# Medallion demo -- gold 02
Idempotent demo pipeline step.

```python
from pyspark.sql import functions as F
spark.sql("CREATE SCHEMA IF NOT EXISTS demo_gold")
df = spark.table("demo_silver.step_02_clean")
metric = df.agg(F.lit(2).alias("metric_step"), F.count("*").alias("record_count"), F.sum("amount").alias("total_amount"), F.avg("amount").alias("average_amount"))
metric.write.mode("overwrite").format("delta").saveAsTable("demo_gold.step_02_metrics")
metric.show()
print("gold step 02 complete")
```
