# Medallion demo -- bronze 03
Idempotent demo pipeline step.

```python
from pyspark.sql import functions as F
spark.sql("CREATE SCHEMA IF NOT EXISTS demo_bronze")
rows = [(301, "item_3_a", 30.0), (302, "item_3_b", 60.0)]
df = spark.createDataFrame(rows, ["record_id", "item", "amount"])
df = df.withColumn("source_step", F.lit(3)).withColumn("ingested_at", F.current_timestamp())
df.write.mode("overwrite").format("delta").saveAsTable("demo_bronze.step_03_raw")
print("bronze step 03 complete")
```
