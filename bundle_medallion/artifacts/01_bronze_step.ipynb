# Medallion demo -- bronze 01
Idempotent demo pipeline step.

```python
from pyspark.sql import functions as F
spark.sql("CREATE SCHEMA IF NOT EXISTS demo_bronze")
rows = [(101, "item_1_a", 10.0), (102, "item_1_b", 20.0)]
df = spark.createDataFrame(rows, ["record_id", "item", "amount"])
df = df.withColumn("source_step", F.lit(1)).withColumn("ingested_at", F.current_timestamp())
df.write.mode("overwrite").format("delta").saveAsTable("demo_bronze.step_01_raw")
print("bronze step 01 complete")
```
