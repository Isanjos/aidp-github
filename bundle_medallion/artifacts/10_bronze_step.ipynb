# Medallion demo -- bronze 10
Idempotent demo pipeline step.

```python
from pyspark.sql import functions as F
spark.sql("CREATE SCHEMA IF NOT EXISTS demo_bronze")
rows = [(1001, "item_10_a", 100.0), (1002, "item_10_b", 200.0)]
df = spark.createDataFrame(rows, ["record_id", "item", "amount"])
df = df.withColumn("source_step", F.lit(10)).withColumn("ingested_at", F.current_timestamp())
df.write.mode("overwrite").format("delta").saveAsTable("demo_bronze.step_10_raw")
print("bronze step 10 complete")
```
