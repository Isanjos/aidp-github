# Medallion demo -- bronze 05
Idempotent demo pipeline step.

```python
from pyspark.sql import functions as F
spark.sql("CREATE SCHEMA IF NOT EXISTS demo_bronze")
rows = [(501, "item_5_a", 50.0), (502, "item_5_b", 100.0)]
df = spark.createDataFrame(rows, ["record_id", "item", "amount"])
df = df.withColumn("source_step", F.lit(5)).withColumn("ingested_at", F.current_timestamp())
df.write.mode("overwrite").format("delta").saveAsTable("demo_bronze.step_05_raw")
print("bronze step 05 complete")
```
