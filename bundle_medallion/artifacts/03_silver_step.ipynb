# Medallion demo -- silver 03
Idempotent demo pipeline step.

```python
from pyspark.sql import functions as F
spark.sql("CREATE SCHEMA IF NOT EXISTS demo_silver")
df = spark.table("demo_bronze.step_03_raw")
clean = df.filter(F.col("amount").isNotNull()).select(F.col("record_id").cast("long"), F.trim("item").alias("item"), F.col("amount").cast("decimal(12,2)").alias("amount"), "source_step", "ingested_at")
clean.write.mode("overwrite").format("delta").saveAsTable("demo_silver.step_03_clean")
print("silver step 03 complete")
```
